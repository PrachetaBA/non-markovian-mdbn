# Visualize CPTs for the DBNs for pooling and parameter extrapolation

This notebook is used to visualize CPTs for DBNs with and without pooling as well as with and without parameter extrapolation. We use these figures in the journal paper. 

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import pyAgrum as gm
import pyAgrum.lib.dynamicBN as gdynbn
```

```python
# Load the relevant DBN
dbn_file = '../data/dbns/dbn_indicator-False_nreps-10_pooled_extrapolation-False.bif'
dbn = gm.BayesNet()
dbn.loadBIF(dbn_file)
```

```python
# Extract the CPT
lambda_val = '_0_2'
mu_val = '1'
lprev = 0
max_ql = 10
r_val = '_0_2'
for lq0 in range(0, max_ql+1):
    probs = dbn.cpt('Lqat')[{'Lambdaqat': lambda_val, 'Muqat': mu_val, 'Lambdaqct': lambda_val, 'Muqct': mu_val, 'Rcat': r_val, 'Lqc0': lprev, 'Lqa0': lq0}]
    plt.plot(probs, label=r'$\ell = $' + str(lq0)) #label=r'$L^{A(j)}$ = ' + str(lq0))
    plt.scatter(range(0, len(probs)), probs)
plt.legend(loc='upper left', bbox_to_anchor=(1.05, 1))
plt.xlabel(r'$k$')
plt.ylabel(r'$P(k)$')
plt.title(r'$P(L^{A(j+1)} = k | L^{A(j)} = \ell, \lambda^{A(j)} = 0.2, \mu^{A(j)} = 1.0, \lambda^{C(j)} = 0.2, \mu^{C(j)} = 1.0, r^{CA(j)} = 0.2, L^{C(j)} = 0)$')
plt.savefig('../figures/journal_figures/without_param_extrapolation.png', bbox_inches='tight', dpi=300)
```

```python
# Load the relevant DBN
dbn_file = '../data/dbns/dbn_indicator-False_nreps-10_pooled_extrapolation-True.bif'
dbn = gm.BayesNet()
dbn.loadBIF(dbn_file)
```

```python
# Extract the CPT
lambda_val = '_0_2'
mu_val = '1'
lprev = 0
max_ql = 10
r_val = '_0_2'
for lq0 in range(0, max_ql+1):
    probs = dbn.cpt('Lqat')[{'Lambdaqat': lambda_val, 'Muqat': mu_val, 'Lambdaqct': lambda_val, 'Muqct': mu_val, 'Rcat': r_val, 'Lqc0': lprev, 'Lqa0': lq0}]
    plt.plot(probs, label=r'$\ell = $' + str(lq0))    # label=r'$L^{A(j)}$ = ' + str(lq0))
    plt.scatter(range(0, len(probs)), probs)
plt.legend(loc='upper left', bbox_to_anchor=(1.05, 1))
plt.xlabel(r'$k$')
plt.ylabel(r'$P(k)$')
plt.title(r'$P(L^{A(j+1)} = k | L^{A(j)} = \ell, \lambda^{A(j)} = 0.2, \mu^{A(j)} = 1.0, \lambda^{C(j)} = 0.2, \mu^{C(j)} = 1.0, r^{CA(j)} = 0.2, L^{C(j)} = 0)$')
plt.savefig('../figures/journal_figures/with_param_extrapolation.png', bbox_inches='tight', dpi=300)
```
