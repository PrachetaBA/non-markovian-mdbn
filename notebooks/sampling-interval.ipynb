# Determine the sampling interval that is appropriate for a given set of simulation input parameters 

This code is used to determine a sampling interval that is appropriate for a given set of simulation input parameters. In general, we use the algorithm that looks at all the exogenous arrival rates and computes a Poisson distribution with parameter $\lambda \delta$ such that $P(n > 1) < \epsilon$ where $n$ is the number of events that occur in a time interval $\delta$. 

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng()
```

```python
# Draw the PDF of a Poisson distribution with parameter lambda * delta 
lambda_ = 0.7
delta = 0.05
# Generate the Poisson distribution for x = 0, 1, 2, ..., 10
x = [i for i in range(11)]
# Generate the Poisson distribution
P = [np.exp(-lambda_ * delta) * (lambda_ * delta) ** i / np.math.factorial(i) for i in x]
# Plot the Poisson distribution
plt.bar(x, P)
plt.xlabel('x')
plt.ylabel('P(x)')
```

```python
# Convert P into a dictionary for easy access
P_dict = {}
for i in range(len(x)):
    P_dict[x[i]] = P[i]
    
# Print the value of P(x) for x = 0, 1, 2, ..., 10
print('P(x):')
for i in range(len(x)):
    print('P(', x[i], ') =', P[i])
```
