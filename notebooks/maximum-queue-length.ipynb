# Find the maximum queue length from the time series generated by the simulator 

This notebook explores the maximum queue length for both the time series generated by the simulator and also the sub-sampled time series. For now, we have it for the experiment_number = 1 from `simulator.yaml`

```python
# Import libraries
import pandas as pd
```

```python
# Read in the time series data
df = pd.read_csv('../data/simulation/time-series-exp-1.csv')
print(f'Columns of the data: {df.columns}')
```

```python
# Find the maximum value among the columns 'QueueAL', 'QueueBL', 'QueueCL'
max_queue = df[['QueueAL', 'QueueBL', 'QueueCL']].max().max()
print(f'Maximum queue length among the columns QueueAL, QueueBL, QueueCL: {max_queue}')
```

```python
# Find the maximum queue length for sub-sampled time series data for each sampling interval
def return_max_ql(expt_number):
    df = pd.read_csv(f'../data/discrete_time/discrete-time-2tbn-exp-{expt_number}.csv')
    max_ql = df[['L_qa_tprev', 'L_qb_tprev', 'L_qc_tprev', 'L_qa', 'L_qb', 'L_qc']].max().max()
    print(f'Maximum queue length for experiment {expt_number}: {max_ql}')
```

```python
return_max_ql(7)
```

```python
return_max_ql(6)
```

```python
return_max_ql(5)
```

```python
return_max_ql(4)
```

```python
return_max_ql(3)
```

```python
return_max_ql(2)
```
